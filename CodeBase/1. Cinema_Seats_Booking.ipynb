````python
from mesa import Agent, Model
from mesa.time import RandomActivation
from mesa.space import MultiGrid
from mesa.datacollection import DataCollector
import numpy as np
````

````python
class CinemaModel(Model):
    def __init__(self, N, width, height):
        self.num_agents = N
        self.grid = MultiGrid(width, height, True)
        self.schedule = RandomActivation(self)

        # Create agents
        for i in range(self.num_agents):
            a = CustomerAgent(i, self)
            x = self.random.randrange(self.grid.width)
            y = self.random.randrange(self.grid.height)
            self.grid.place_agent(a, (x, y))
            self.schedule.add(a)

        # Data collector
        self.datacollector = DataCollector(
            agent_reporters={"Booking": "booking"}
        )

    def step(self):
        self.datacollector.collect(self)
        self.schedule.step()
````

# Simulation Grid

A simulation grid is a spatial representation used in agent-based modeling and other types of simulations to define the environment in which agents or entities interact. Let me break this down further:

1. Structure:
   - The grid is typically a 2D or 3D space divided into discrete cells or locations.
   - In the code you showed, it's a 2D grid with a specified width and height.

2. Purpose:
   - It provides a structure for positioning and moving agents within the simulation.
   - It can represent physical space (like a cinema floor) or more abstract spaces.

3. Properties:
   - Cells in the grid can have properties or contain objects.
   - In some simulations, multiple agents can occupy the same cell (as suggested by the use of MultiGrid in your code).

4. Boundaries:
   - Grids can have different boundary conditions. In your code, the `True` parameter in `MultiGrid(width, height, True)` likely indicates that the grid "wraps around," meaning an agent moving off one edge appears on the opposite edge.

5. Movement and Interaction:
   - Agents can move between cells on the grid.
   - The grid structure often determines which other agents or objects an agent can interact with (e.g., nearest neighbors).

6. Visualization:
   - Grids are often easy to visualize, making it simple to display the state of the simulation at any given time.

In the context of your cinema simulation, the grid might represent the physical layout of the cinema, with each cell possibly representing a seat, a section of the lobby, or some other relevant space within the cinema. The agents (customers) are placed and can move around on this grid as the simulation progresses.

This type of spatial representation is particularly useful for simulations where the physical positioning of agents is important to their behavior or interactions. It provides a structured way to model space in discrete simulations.

````python
class CustomerAgent(Agent):
    """ An agent with random booking behavior. """
    def __init__(self, unique_id, model):
        super().__init__(unique_id, model)
        self.booking = 0

    def step(self):
        x = self.random.randrange(self.model.grid.width)
        y = self.random.randrange(self.model.grid.height)

        # Try to book a seat
        cell_content = self.model.grid.get_cell_list_contents([(x, y)])
        if len(cell_content) == 0:
            self.booking = 1
            print(f"Customer {self.unique_id} successfully booked seat ({x}, {y})")
        else:
            self.booking = 0
            print(f"Customer {self.unique_id} failed to book seat ({x}, {y})")
````

This class represents a customer in the cinema simulation. During each step, the customer tries to book a random seat in the cinema. If the chosen seat is unoccupied, the booking is successful; otherwise, it fails. The `booking` attribute keeps track of whether the customer has successfully booked a seat (1) or not (0).

This simple model assumes that customers randomly select seats and only attempt to book once per step. In a more complex simulation, you might add more sophisticated behavior, such as customers preferring certain seats or trying multiple times to find an available seat.

````python
width = 10
height = 10
num_agents = 15
````

````python
model = CinemaModel(num_agents, width, height)
````

````python
for i in range(1):
    model.step()
````

````python
# Print final booking results
for agent in model.schedule.agents:
    print(f"Customer {agent.unique_id} booking status: {agent.booking}")
````

````python
for x in range(model.grid.width):
    for y in range(model.grid.height):
        cell_content = model.grid.get_cell_list_contents([(x, y)])
        print(f"Cell ({x}, {y}) has {len(cell_content)} agents")
````
